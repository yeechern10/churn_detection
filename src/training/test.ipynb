```python
import mlflow
from mlflow.tracking import MlflowClient
from configuration import MLFLOW_TRACKING_URI

import pandas as pd

client = MlflowClient()
mlflow.set_tracking_uri(MLFLOW_TRACKING_URI)
exp = client.get_experiment_by_name("churn_detection")
print(f"Experiment ID: {exp.experiment_id}, Name: {exp.name}")
all_runs_df = mlflow.search_logged_models(experiment_ids=[f"{exp.experiment_id}"], order_by=[{"field_name": "metrics.recall", "ascending": False}])
all_runs_df.iloc[0]

```
